```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X=2*np.random.rand(100,1)
y=4+3*X+np.random.randn(100,1)
```

```python
plt.scatter(X,y)
plt.xlabel("X")
plt.ylabel('y')
plt.show()
```

```python
X_b=np.concatenate([np.ones((100,1)),X],axis=1)
```

```python
theta_best=np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
```

```python
theta_best
```

- Now lets make predictions using above equation i.e y=4+3x+Gaussian_noise

```python
X_new=np.array([[0],[2]])
x_new_b=np.concatenate([np.ones((2,1)),X_new],axis=1)
y_predict=x_new_b.dot(theta_best)
y_predict
```

```python
plt.plot(X_new,y_predict,'r-',label='prediction')
plt.plot(X,y,"b.",label='acutal data')
plt.axis([0, 2, 0, 15])
plt.legend()
plt.show()
```

# Learning Curves

######  -  These are plots of the model’s perfor‐mance on the training set and the validation set as a function of the training set size(or the training iteration). To generate the plots, simply train the model several times on different sized subsets of the training set. The following code defines a function that plots the learning curves of a model given some training data

```python
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
def plot_learning_curves(model,X,y):
    X_train,X_val,y_train,y_val=train_test_split(X,y,test_size=0.2)
    train_errors,val_errors=[],[]
    for m in range(1,len(X_train)):
        model.fit(X_train[:m],y_train[:m])
        y_train_predict=model.predict(X_train[:m])
        y_val_predict=model.predict(X_val)
        
        
        train_errors.append(mean_squared_error(y_train[:m],y_train_predict))
        val_errors.append(mean_squared_error(y_val,y_val_predict))
        
    plt.plot(np.sqrt(train_errors),'r-+',label='train',linewidth=2)
    plt.plot(np.sqrt(val_errors),'b-',linewidth=3,label='validation')
    plt.legend()
    plt.xlabel('Training Set size')
    plt.ylabel("RMSE")
```

```python
lin_reg = LinearRegression()
plot_learning_curves(lin_reg, X, y)
```

#### Let's Look at the model performance on the Training Data.
- Initially, when there were one or two instances in the training set,the model fit them perfectly,so the curve started at 0.
- As the instances on the training set increased, the model started to make few mistakes and hence the RMSE increased upto the 1.5 which was at the size of 10/11. This was a  point at which adding new instances to the training set didn't make the average much better or worse.
#### Let's Look at the model performance on the Validation Set.
- When the model was trained on the very few dataset, it was incapable of generalizing the new data points and henece validation error was big.
- As the model was showen up with more instances, the validation error slowely gone down.

###### These learning curves are typical of an underfitting model. Both curves have reached a plateau; they are close and fairly high.


```python
from sklearn.pipeline import Pipeline
```

```python
polynomial_regression = Pipeline([
 ("poly_features", PolynomialFeatures(degree=2, include_bias=False)),
 ("lin_reg", LinearRegression()),
 ])
```

```python
plot_learning_curves(polynomial_regression, X, y)
```

- ##### If our model is underfitting the training data, adding more training examples will not help. We need to use a more complex model or come up with better features.
- ##### One way to improve an overfitting model is to feed it more training data until the validation error reaches the training error.

# The Bias/Variance Tradeoff
- The model `generalization Error` can be explained as the sum of three different errors:

 ####  `Bias:- ` This Part of generalization error is due to wrong assumptions, such as assuming that the data is linear when it is actually quardatic. A high bias model likely underfit the data.
 
 #### `Variance:-` This part is due to the model’s excessive sensitivity to small variations in the training data. A model with many degrees of freedom (such as a high-degree polynomial model) is likely to have high variance, and thus to overfit the training data.
 
 #### `Irreducible Error:-` This part is due to the noisiness of the data itself. The only way to reduce this part of the error is to clean up the data (e.g., fix the data sources, such as broken sensors, or detect and remove outliers).
  
 - Increasing a model’s complexity will typically increase its variance and reduce its bias.Conversely, reducing a model’s complexity increases its bias and reduces its variance. This is why it is called a tradeoff.
 
###### Real Time Analogy with `Dartboard`
 1. High Bias (Underfitting)
     - All the darts land far from the center, but they are grouped together.
     - This means the model makes consistent but wrong predictions (e.g., always predicting 140 cm instead of 160 cm).
     
 2. High Variance (Overfitting)
     - Darts are scattered all over the board, sometimes near the bullseye, sometimes far.
     - The model learns too much noise, so predictions are inconsistent.
 3. Low Bias & Low Variance (Good Model)
     - Darts land close to the center and are grouped tightly.
     - The model makes small and consistent errors.

# Regularized Linear Models
- A good way to reduce overfitting(i.e to constraint it).

### 1. Ridge Regression (Tikhonov Regularization)


- Ridge Regression Equation.<br>
 $$L= MSE({\theta})+{\alpha}\frac{1}{2}\sum_{i=1}^{n}\theta_i^2 $$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline  import make_pipeline
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(42)
X=3*np.random.rand(30,1)
y=1.5+X+0.5*X**2+np.random.randn(30,1)*0.5
X_plot=np.linspace(0,3,100).reshape(-1,1)
```

```python
alphas = [0, 1e-5, 1, 10, 100]
colors = ['b', 'g', 'r', 'purple', 'orange']
linestyles = ['-', '--', ':', '-.', (0, (3, 1, 1, 1))]
```

```python
plt.figure(figsize=(10, 4))

for i,alpha in enumerate(alphas):
    model=make_pipeline(PolynomialFeatures(degree=10),Ridge(alpha=alpha))
    model.fit(X,y)
    y_plot=model.predict(X_plot)
    
    plt.plot(X_plot,y_plot,linestyle=linestyles[i],color=colors[i],label=f'alpha={alpha}')

plt.scatter(X,y,color='blue',s=20,label='Data',zorder=3)
plt.xlabel(r'$x_1$', fontsize=14)
plt.ylabel(r'$y$', fontsize=14)
plt.legend()
plt.title("Ridge Regression with Different α Values")
plt.show()
    
```

# Lasso Regression
- Least Absolute Shrinkage and Selection Operator Regression (simply called LassoRegression) is another regularized version of Linear Regression: just like Ridge Regression, it adds a regularization term to the cost function, but it uses the ℓ1 norm of the weight vector instead of half the square of the ℓ2 norm

#### - Lasso Regression Cost Function:
$$ J(\theta)=MSE(\theta)+{\alpha}\sum_{i=1}^{n}|\theta_i| $$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.preprocessing import StandardScaler
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import make_pipeline

np.random.seed(42)
X = np.linspace(-3, 3, 100).reshape(-1, 1)  
y = X**3 - 3*X**2 + 2 + np.random.normal(0, 3, size=X.shape) 

lambdas = [0.001, 0.01, 0.1, 1, 10]
colors = ['red', 'blue', 'green', 'purple', 'orange']

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='black', label="Original Data", alpha=0.6)

for alpha, color in zip(lambdas, colors):
    model = make_pipeline(PolynomialFeatures(degree=5), StandardScaler(), Lasso(alpha=alpha, max_iter=10000))
    model.fit(X, y)
    y_pred = model.predict(X)

    plt.plot(X, y_pred, color=color, label=f"λ = {alpha}")

# Plot settings
plt.xlabel("Feature X")
plt.ylabel("Target Y")
plt.title("Polynomial Lasso Regression with Different Lambda Values")
plt.legend()
plt.show()
```
